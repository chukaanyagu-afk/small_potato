```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
```

```python
import io
csv_data = """Car,Model,Volume,Weight,CO2
Toyota,Aygo,1000,790,99
Mitsubishi,Space Star,1200,1160,95
Skoda,Citigo,1000,929,95
"""
df = pd.read_csv(io.StringIO(csv_data))
```

```python
x=df[:-1].copy()
x.head()
```

```python
for i in x.columns:
    if pd.api.types.is_numeric_dtype(x[i]):
        x[i]=x[i].fillna(x[i].mean())
for i in x.columns:
    print(x[i].isnull().sum())
```

```python
x=x.iloc[:,:-1].values
print(x)
```

```python
x=df[df.columns[:-1]].copy()
x.head()
```

```python
for i in x.columns:
    if pd.api.types.is_numeric_dtype(x[i]):
        x[i]=x[i].fillna(x[i].mean())
for i in x.columns:
    print(x[i].isnull().sum())
```

```python
from sklearn.cluster import KMeans
```

```python
import numpy as np
wcss=[]
n_samples = x.select_dtypes(include=np.number).shape[0]
for i in range(1, min(11, n_samples + 1)): # Limit the range to not exceed n_samples
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=0)
    # Select only numeric columns from x for KMeans
    x_numeric = x.select_dtypes(include=np.number)
    kmeans.fit(x_numeric)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, len(wcss) + 1), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
import numpy as np
kmeans=KMeans(n_clusters=2,init='k-means++',max_iter=300,n_init=10,random_state=0)
y_kmeans=kmeans.fit_predict(x.select_dtypes(include=np.number))
```

```python
print(y_kmeans)
```

```python
plt.scatter(x_numeric.iloc[y_kmeans==0,0],x_numeric.iloc[y_kmeans==0,1],s=100,c='red',label='Cluster 1')
plt.scatter(x_numeric.iloc[y_kmeans==1,0],x_numeric.iloc[y_kmeans==1,1],s=100,c='blue',label='Cluster 2')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],s=300,c='yellow',label='Centroids')
plt.title('Clusters of cars')
plt.xlabel('Volume')
plt.ylabel('Weight')
plt.legend()
plt.show()
```
